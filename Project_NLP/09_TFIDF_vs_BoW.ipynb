```python
# TF-IDF implementation and comparison with BoW
```

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

docs = ["cats like milk", "cats like fish", "dogs like bones"]
bow = CountVectorizer()
tfidf = TfidfVectorizer()

print("Vocabulary:", bow.fit(docs).get_feature_names_out())
print("BoW:\n", bow.fit_transform(docs).toarray())
print("TF-IDF:\n", tfidf.fit_transform(docs).toarray().round(3))
```
