```python
# TF-IDF implementation and comparison with BoW
```

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
corpus = ['cats are cute', 'cats are smart', 'dogs are cute']
bow = CountVectorizer().fit_transform(corpus)
tfidf = TfidfVectorizer().fit_transform(corpus)
print('BoW:')
print(bow.toarray())
print('TF-IDF:')
print(tfidf.toarray())
```
