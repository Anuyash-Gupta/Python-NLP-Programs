```python
import nltk
import spacy
from nltk.tokenize import sent_tokenize, word_tokenize

nltk.download('punkt')
nltk.download('punkt_tab')

text = "Natural Language Processing is a branch of Artificial Intelligence. It helps computers understand human language."

print("NLTK Sentence Tokenization:")
print(sent_tokenize(text))

print("\nNLTK Word Tokenization:")
print(word_tokenize(text))

nlp = spacy.load("en_core_web_sm")
doc = nlp(text)

print("\nspaCy Sentence Tokenization:")
for sentence in doc.sents:
    print(sentence.text)

print("\nspaCy Word Tokenization:")
print([token.text for token in doc])
```
