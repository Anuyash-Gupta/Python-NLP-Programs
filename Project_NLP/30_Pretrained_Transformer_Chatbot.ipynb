```python
# Chatbot using a pretrained DialoGPT model
```

```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

name = "microsoft/DialoGPT-small"
tokenizer = AutoTokenizer.from_pretrained(name)
model = AutoModelForCausalLM.from_pretrained(name)
chat_history = None

print("Chatbot ready. Type quit to stop.")
while True:
    user_text = input("You: ")
    if user_text.strip().lower() == "quit":
        break
    new_input = tokenizer.encode(user_text + tokenizer.eos_token, return_tensors="pt")
    bot_input = torch.cat([chat_history, new_input], dim=-1) if chat_history is not None else new_input
    chat_history = model.generate(bot_input, max_length=min(bot_input.shape[-1] + 80, 1000),
                                  pad_token_id=tokenizer.eos_token_id,
                                  do_sample=True, top_p=0.9, temperature=0.75)
    reply = tokenizer.decode(chat_history[:, bot_input.shape[-1]:][0], skip_special_tokens=True)
    print("Bot:", reply)
```
