```python
# Word2Vec using Gensim on a custom corpus
```

```python
from gensim.models import Word2Vec

sentences = [
    ["nlp", "uses", "machine", "learning"],
    ["machine", "learning", "uses", "data"],
    ["language", "models", "learn", "words"],
    ["nlp", "models", "process", "language"],
    ["data", "helps", "machine", "learning"]
]
model = Word2Vec(sentences, vector_size=50, window=3, min_count=1, workers=1, sg=1, epochs=100)
print("Vector for machine:", model.wv["machine"][:10])
print("Similar words:", model.wv.most_similar("machine", topn=3))
```
