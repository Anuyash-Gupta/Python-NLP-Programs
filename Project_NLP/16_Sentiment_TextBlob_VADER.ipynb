```python
# Sentiment analysis using TextBlob and VADER
```

```python
from textblob import TextBlob
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

samples = ["This product is wonderful!", "The service was terrible.", "It is okay, nothing special."]
vader = SentimentIntensityAnalyzer()

for sentence in samples:
    print("\nText:", sentence)
    print("TextBlob polarity:", round(TextBlob(sentence).sentiment.polarity, 3))
    print("VADER:", vader.polarity_scores(sentence))
```
