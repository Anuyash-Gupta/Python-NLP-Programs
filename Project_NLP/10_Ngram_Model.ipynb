```python
# Unigram, bigram, and trigram generation
```

```python
from nltk.util import ngrams
from nltk.tokenize import wordpunct_tokenize

corpus = "Natural language processing helps computers understand human language."
tokens = wordpunct_tokenize(corpus.lower())

for n, label in [(1, "Unigrams"), (2, "Bigrams"), (3, "Trigrams")]:
    print(label, list(ngrams(tokens, n)))
```
