```python
import scipy.linalg as sl
if not hasattr(sl, 'triu'):
    import numpy as np
    sl.triu = np.triu
from gensim.models import Word2Vec
sentences = [['nlp','uses','text'], ['word','embeddings','represent','words'], ['nlp','models','learn','patterns'], ['words','have','vectors']]
model = Word2Vec(sentences, vector_size=20, window=3, min_count=1, workers=1, seed=42, epochs=30)
print('Vector for nlp:', model.wv['nlp'][:8])
print('Similar words:', model.wv.most_similar('nlp', topn=2))
```
