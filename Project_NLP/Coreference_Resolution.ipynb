```python
# Coreference Resolution (spaCy / neuralcoref-style)
```

```python
text = "Alice went to the library. She borrowed a book because it was interesting."
                                                                
                                                                                     
antecedent = "Alice"
resolved = text.replace("She", antecedent).replace("she", antecedent).replace("it", "the book")
print("Original:", text)
print("Heuristic resolution:", resolved)
```
