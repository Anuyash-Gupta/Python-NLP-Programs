```python
from pathlib import Path
import numpy as np

GLOVE_PATH = Path("glove.6B.50d.txt")                                                
embeddings = {}

if GLOVE_PATH.exists():
    with GLOVE_PATH.open(encoding="utf8") as f:
        for line in f:
            parts = line.rstrip().split()
            word, values = parts[0], parts[1:]
            embeddings[word] = np.asarray(values, dtype="float32")
    print("Loaded", len(embeddings), "vectors")
    for word in ["king", "queen", "computer"]:
        if word in embeddings:
            print(word, embeddings[word][:8])
else:
    print("GloVe file not found. Download a GloVe text embedding file and set GLOVE_PATH.")
```
