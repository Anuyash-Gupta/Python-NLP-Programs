```python
# Hybrid probabilistic + FSM document boundary detection
```

```python
import re

text = (
    "ABSTRACT: This paper studies language models. "
    "METHODS: We collected text and trained a model. "
    "RESULTS: Accuracy improved. "
    "CONCLUSION: The approach is useful."
)
sentences = re.split(r"(?<=[.!?])\s+", text.strip())
states = {"ABSTRACT": "ABSTRACT", "METHODS": "METHODS",
          "RESULTS": "RESULTS", "CONCLUSION": "CONCLUSION"}
current_state = "BODY"
sections = {}

for sentence in sentences:
                                                                                               
    match = re.match(r"^(ABSTRACT|METHODS|RESULTS|CONCLUSION):\s*(.*)", sentence)
    boundary_score = 0.99 if match else 0.05
    if boundary_score >= 0.8 and match:
        current_state = states[match.group(1)]
        content = match.group(2)
    else:
        content = sentence
    sections.setdefault(current_state, []).append(content)

for section, content in sections.items():
    print(f"\n[{section}]")
    print(" ".join(content))
```
