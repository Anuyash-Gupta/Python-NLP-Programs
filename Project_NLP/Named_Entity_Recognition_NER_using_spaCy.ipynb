```python
import spacy

nlp = spacy.load("en_core_web_sm")

text = "Sundar Pichai is the CEO of Google. Google is headquartered in Mountain View, California. He was born in Chennai, India."

doc = nlp(text)

print("Named Entities:")
for entity in doc.ents:
    print(f"{entity.text:25} -> {entity.label_} -> {spacy.explain(entity.label_)}")
```
