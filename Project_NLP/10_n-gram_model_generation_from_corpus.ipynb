```python
# N-Gram model generation from corpus
```

```python
from sklearn.feature_extraction.text import CountVectorizer
corpus = ['natural language processing is fun', 'language processing uses models']
for n in (1, 2, 3):
    vec = CountVectorizer(ngram_range=(n,n))
    mat = vec.fit_transform(corpus)
    print(f'{n}-gram:', vec.get_feature_names_out().tolist())
    print(mat.toarray())
```
