```python
# Topic modeling using Latent Semantic Analysis
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD

docs = [
 "football team scored goal",
 "player trains for football",
 "computer data machine learning",
 "machine learning model data",
 "football team won match"
]
X = TfidfVectorizer(stop_words="english").fit_transform(docs)
lsa = TruncatedSVD(n_components=2, random_state=42)
doc_topics = lsa.fit_transform(X)
print("Document-topic representation:\n", doc_topics.round(3))
print("Explained variance ratio:", lsa.explained_variance_ratio_.round(3))
```
