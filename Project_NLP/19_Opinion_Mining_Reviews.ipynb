```python
# Opinion mining on product/service reviews
```

```python
from textblob import TextBlob

reviews = [
 "The camera quality is excellent but the battery is weak.",
 "Delivery was fast and customer support was helpful.",
 "The screen is beautiful, however the price is too high."
]
aspects = ["camera", "battery", "delivery", "support", "screen", "price"]

for review in reviews:
    polarity = TextBlob(review).sentiment.polarity
    mentioned = [aspect for aspect in aspects if aspect in review.lower()]
    label = "positive" if polarity > 0.1 else "negative" if polarity < -0.1 else "neutral"
    print({"review": review, "aspects": mentioned, "polarity": round(polarity, 3), "sentiment": label})
```
