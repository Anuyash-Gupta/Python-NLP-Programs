```python
# Text classification using Naïve Bayes and SVM with TF-IDF
```

```python
from sklearn.pipeline import make_pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import LinearSVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report

texts = [
 "win money now", "claim your free prize", "limited offer click now",
 "team meeting tomorrow", "please send the report", "project update attached",
 "free cash reward", "let us discuss the assignment", "your account prize",
 "can we meet after class", "exclusive offer today", "notes for the lecture"
]
labels = ["spam","spam","spam","ham","ham","ham","spam","ham","spam","ham","spam","ham"]
X_train, X_test, y_train, y_test = train_test_split(texts, labels, test_size=0.25, random_state=42, stratify=labels)

for name, clf in [("Naive Bayes", MultinomialNB()), ("Linear SVM", LinearSVC())]:
    model = make_pipeline(TfidfVectorizer(), clf)
    model.fit(X_train, y_train)
    print("\n", name)
    print(classification_report(y_test, model.predict(X_test), zero_division=0))
```
