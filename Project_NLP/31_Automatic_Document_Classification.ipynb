```python
# Automatic document classification and separation
```

```python
from pathlib import Path
from sklearn.pipeline import make_pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC

train_texts = [
 "football match team goal", "player scored a goal",
 "python code software development", "computer programming language",
 "hospital doctor patient health", "medicine and clinical care"
]
train_labels = ["sports", "sports", "technology", "technology", "health", "health"]
classifier = make_pipeline(TfidfVectorizer(), LinearSVC())
classifier.fit(train_texts, train_labels)

documents = [
 "the team won the football match",
 "write code in python",
 "doctor treats the patient"
]
output_dir = Path("classified_documents")
for text in documents:
    category = classifier.predict([text])[0]
    folder = output_dir / category
    folder.mkdir(parents=True, exist_ok=True)
    filename = f"document_{documents.index(text)+1}.txt"
    (folder / filename).write_text(text, encoding="utf-8")
    print(f"{filename} -> {category}")
```
