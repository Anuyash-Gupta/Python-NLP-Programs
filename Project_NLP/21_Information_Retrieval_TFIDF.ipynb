```python
# Information retrieval with TF-IDF ranking
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

documents = [
 "Python is a programming language used for data science",
 "Java is used for enterprise application development",
 "Machine learning uses data and statistical models",
 "Natural language processing works with text"
]
query = "data science and machine learning"
vectorizer = TfidfVectorizer()
matrix = vectorizer.fit_transform(documents + [query])
scores = cosine_similarity(matrix[-1], matrix[:-1]).ravel()
for idx in scores.argsort()[::-1]:
    print(f"{scores[idx]:.3f} | {documents[idx]}")
```
