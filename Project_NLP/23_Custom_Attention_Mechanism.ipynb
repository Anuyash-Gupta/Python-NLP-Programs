```python
# Implement a basic custom attention mechanism
```

```python
import numpy as np

                                                                  
H = np.array([
    [0.2, 0.1, 0.7],
    [0.8, 0.2, 0.1],
    [0.1, 0.9, 0.3]
], dtype=float)
query = np.array([0.5, 0.2, 0.4])

scores = H @ query                                                  
exp_scores = np.exp(scores - scores.max())
weights = exp_scores / exp_scores.sum()
context = weights @ H

print("Scores:", scores.round(3))
print("Attention weights:", weights.round(3))
print("Context vector:", context.round(3))
```
