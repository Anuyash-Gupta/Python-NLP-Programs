```python
# Topic modeling using Latent Dirichlet Allocation
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import LatentDirichletAllocation

docs = [
 "football team scored goal match",
 "player training football team",
 "computer software data machine learning",
 "machine learning trains computer models",
 "team won football championship",
 "data science uses computer software"
]
vectorizer = CountVectorizer(stop_words="english")
X = vectorizer.fit_transform(docs)
lda = LatentDirichletAllocation(n_components=2, random_state=42, max_iter=10)
lda.fit(X)
terms = vectorizer.get_feature_names_out()

for topic_id, weights in enumerate(lda.components_):
    top = weights.argsort()[-5:][::-1]
    print(f"Topic {topic_id+1}:", [terms[i] for i in top])
```
