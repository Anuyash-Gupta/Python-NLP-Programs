```python
# Text similarity using Word Mover’s Distance (WMD)
```

```python
from gensim.models import KeyedVectors

                                                                                 
VECTOR_PATH = "GoogleNews-vectors-negative300.bin"
try:
    vectors = KeyedVectors.load_word2vec_format(VECTOR_PATH, binary=True)
    a = "machine learning is useful".lower().split()
    b = "computers learn patterns".lower().split()
    print("WMD:", vectors.wmdistance(a, b))
except FileNotFoundError:
    print(f"Vector file not found: {VECTOR_PATH}. Set VECTOR_PATH to your pretrained vectors.")
```
