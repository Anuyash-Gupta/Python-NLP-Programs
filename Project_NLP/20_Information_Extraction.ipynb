```python
import re

text = "Contact Riya at riya@example.com. The meeting is on 12/11/2026."
emails = re.findall(r"[\w.+-]+@[\w.-]+\.[A-Za-z]{2,}", text)
dates = re.findall(r"\b\d{1,2}/\d{1,2}/\d{4}\b", text)
print("Emails:", emails)
print("Dates:", dates)

                         
                    
                                          
try:
    import spacy
    nlp = spacy.load("en_core_web_sm")
    doc = nlp(text)
    print("Entities:", [(ent.text, ent.label_) for ent in doc.ents])
except Exception as e:
    print("spaCy model not available; install en_core_web_sm to run NER.", e)
```
